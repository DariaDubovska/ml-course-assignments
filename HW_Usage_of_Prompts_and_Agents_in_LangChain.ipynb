````python
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_openai import ChatOpenAI
from langchain_classic.agents import load_tools, Tool, AgentExecutor, AgentType, create_react_agent, initialize_agent
````

````python
tavily_tool = TavilySearchResults(max_results=5)
tools = [tavily_tool]

llm = ChatOpenAI(model="o3-mini")
prompt_template = """Answer the following questions as best you can. You have access to the following tools:

{tools}

Use the following format:

Question: the input question you must answer
Thought: you should always think about what to do
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
... (this Thought/Action/Action Input/Observation can repeat N times)
Thought: I now know the final answer
Final Answer: the final response to the original input question

Begin!

Question: Find the 5 most recent scientific publications on the topic of "{topic}".
IMPORTANT INSTRUCTIONS:
1. You MUST find 5 REAL and SPECIFIC publications.
2. If the initial search results are incomplete or missing authors/descriptions, you MUST perform additional targeted searches for each paper title found.
3. DO NOT provide "Example" answers, "Placeholders", or general advice on how to find papers.
4. Provide ONLY actual data found during your search actions.
5. For each publication provide: Title, Authors, and a 1-2 sentence Description.

Make sure the publications are real and recently published.

Thought: {agent_scratchpad}"""
````

````python
prompt = PromptTemplate(

    template=prompt_template,
    input_variables=["tools", "tool_names", "topic", "agent_scratchpad"]
)
````

````python
agent = create_react_agent(llm, tools, prompt=prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)
````

````python
topic = "Artificial intelligence"
agent_executor.invoke({"topic": topic})
````

````python
topic = "Biotechnology"
agent_executor.invoke({"topic": topic})
````

````python
from langchain_experimental.utilities import PythonREPL
````

````python
python_repl = PythonREPL()
python_tool = Tool(
    name="python_interpreter",
    func=python_repl.run,
    description="""Use this to perform mathematical calculations and data analysis.
    IMPORTANT: Provide ONLY clean Python code.
    Do NOT include decorative lines (like '---'), markdown code blocks, or comments inside the action input."""
)

tools.append(python_tool)
````

````python
prompt_template = """
You are a Professional Business Analyst and Strategic Consultant. Your goal is to provide a precise export forecast based on historical data, weather patterns, and global economic trends.

You have access to the following tools:
{tools}

To use a tool, please use the following format:
Thought: Do I need to use a tool? Yes
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action

When you have a response for the user, or if you do not need to use a tool, you MUST use the format:
Thought: I now know the final answer
Final Answer: [your response here]

IMPORTANT RULES:
1. You MUST use the search tool to verify weather and economic data for 2026. Internal knowledge is not enough.
2. You are FORBIDDEN from providing a Final Answer without using BOTH tools at least once.

Follow this logical workflow:
1. **Data Extraction**: Extract the historical export volumes and years from the user's text.
2. **Mathematical Trend**: You MUST use the "python_interpreter" for any and all calculations required.
3. **External Research**: Use the search tool to find current information about:
   - Weather conditions in Brazil for the 2026 orange harvest (e.g., droughts, frosts, or El Niño impact).
   - Global demand for oranges and current economic factors affecting the citrus market in 2026.
4. **Final Evaluation**: Combine the mathematical projection with the real-world findings. Adjust the forecast up or down based on the "Observation" from your research.

Output Format:
- **Historical Analysis Summary**: Brief recap of the data.
- **Mathematical Projection**: The result from the Python calculation.
- **Market & Weather Insights**: Key findings from the search.
- **Final Forecast for 2026**: A specific estimated tonnage with a brief business justification.

Question: {input}

Thought: {agent_scratchpad}
"""
````

````python
prompt = PromptTemplate(

    template=prompt_template,
    input_variables=["tools", "tool_names","input", "agent_scratchpad"]
)
````

````python
agent = create_react_agent(llm, tools, prompt=prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)
````

````python
user_task = "We export oranges from Brazil. In 2022 exported 200t, in 2023 - 190t, in 2024 - 210t, in 2025 - 220t. Make an estimate of how many oranges we will be able to export in 2026 considering the weather conditions in Brazil and the demand for oranges in the world based on the economic situation."

agent_executor.invoke({"input": user_task})
````
